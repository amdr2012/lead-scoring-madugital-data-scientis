```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Memuat dataset (ganti path sesuai lokasi file Anda atau gunakan URL CSV)
df = pd.read_csv(url)

# Menampilkan informasi awal dataset
print("Ukuran Dataset:", df.shape)
print("\nKolom dalam dataset:")
print(df.columns.tolist())

# Melihat distribusi target (Converted: 1 = Membeli, 0 = Tidak)
print("\nDistribusi Target (Converted):")
print(df['Converted'].value_counts(normalize=True))
```

```python
# 1. Mengganti string 'Select' menjadi NaN agar mudah dideteksi sebagai missing value
df = df.replace('Select', np.nan)

# 2. Menghapus kolom unik atau ID yang tidak berpengaruh pada prediksi
cols_to_drop = ['Prospect ID', 'Lead Number']
df = df.drop(columns=cols_to_drop, errors='ignore')

# 3. Menghapus kolom dengan persentase missing value terlalu tinggi (> 40%)
missing_threshold = 0.4
high_missing_cols = df.columns[df.isnull().mean() > missing_threshold]
df = df.drop(columns=high_missing_cols)

print("Kolom setelah pembersihan awal:", df.shape[1])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# Memisahkan fitur (X) dan target (y)
X = df.drop(columns=['Converted'])
y = df['Converted']

# Memisahkan kolom numerik dan kategorikal
num_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()
cat_cols = X.select_dtypes(include=['object', 'category']).columns.tolist()

# Imputasi Missing Values
# Untuk numerik isi dengan median, untuk kategorikal isi dengan modus ('Missing')
for col in num_cols:
    X[col] = X[col].fillna(X[col].median())

for col in cat_cols:
    X[col] = X[col].fillna('Missing')

# Melakukan One-Hot Encoding pada fitur kategorikal
X_encoded = pd.get_dummies(X, columns=cat_cols, drop_first=True)

# Membagi data menjadi Training set dan Testing set (80:20)
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42, stratify=y)

# Melakukan Scaling pada fitur numerik
scaler = StandardScaler()
X_train[num_cols] = scaler.fit_transform(X_train[num_cols])
X_test[num_cols] = scaler.transform(X_test[num_cols])
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, roc_auc_score, roc_curve

# Melatih model Random Forest
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# Prediksi probabilitas pada data uji
y_pred_proba = rf_model.predict_proba(X_test)[:, 1]
y_pred = rf_model.predict(X_test)

# Evaluasi Performa Model
print("ROC-AUC Score:", roc_auc_score(y_test, y_pred_proba))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
# Menghitung skor lead (probabilitas dikali 100)
df_test_results = X_test.copy()
df_test_results['Actual_Converted'] = y_test.values
df_test_results['Conversion_Probability'] = y_pred_proba
df_test_results['Lead_Score'] = (y_pred_proba * 100).round(2)

# Mengurutkan lead dari skor tertinggi (Hot Leads) ke terendah
ranked_leads = df_test_results.sort_values(by='Lead_Score', ascending=False)

# Menampilkan 10 lead teratas yang paling potensial dihubungi
print(ranked_leads[['Lead_Score', 'Conversion_Probability']].head(10))
```

```python
# Mendapatkan tingkat pentingnya fitur dari Random Forest
importances = rf_model.feature_importances_
feature_imp = pd.DataFrame({'Feature': X_encoded.columns, 'Importance': importances})
feature_imp = feature_imp.sort_values(by='Importance', ascending=False)

# Menampilkan 10 fitur teratas
print(feature_imp.head(10))
```
